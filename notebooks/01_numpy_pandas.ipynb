## Data Cleaning

```python
students = {
    "Name": ["Ali", "Sara", "John", "Aisha", "David", "Ali"],
    "Age": [19, 20, np.nan, 19, 21, 19],
    "Marks": [78, 91, 45, np.nan, 67, 78],
    "Study_Hours": [3, 5, 2, 4, np.nan, 3]
}

df_dirty = pd.DataFrame(students)
```

```python
print(df_dirty)
```

```python
cleaned_df=df_dirty.dropna()
print(cleaned_df)
```

```python
filled_df= df_dirty.copy()

filled_df["Marks"]= filled_df["Marks"].fillna(
    filled_df["Marks"].median()
)
print(filled_df)
```

```python
filled_df= df_dirty.copy()

filled_df["Age"]= filled_df["Age"].fillna(
    filled_df["Age"].mean()
)
print(filled_df)
```

```python
df_dirty.duplicated().sum()
```

```python
deduplicated_df=df_dirty.drop_duplicates()
print(deduplicated_df)
```

```python
cleaned_df = df_dirty.copy()

# Fill missing numerical values
cleaned_df["Marks"] = cleaned_df["Marks"].fillna(
    cleaned_df["Marks"].median()
)

cleaned_df["Age"] = cleaned_df["Age"].fillna(
    cleaned_df["Age"].mean()
)

cleaned_df["Study_Hours"] = cleaned_df["Study_Hours"].fillna(
    cleaned_df["Study_Hours"].median()
)

# Remove duplicate rows
cleaned_df = cleaned_df.drop_duplicates()

# Verify
print(cleaned_df)
print(cleaned_df.isnull().sum())
print("Total missing values:", cleaned_df.isnull().sum().sum())
print("Total duplicate rows:", cleaned_df.duplicated().sum())
```

## Data Types and String Cleaning

```python
validation_df = pd.DataFrame({
    "Name": ["Ali", "Sara", "John", "Aisha", "David"],
    "Age": [19, -5, 18, 19, 150],
    "Marks": [78, 91, 150, -10, 67]
})

print(validation_df)
```

```python
invalid_age= validation_df[
    (validation_df["Age"]<16)|
    (validation_df["Age"]>30)
]

print(invalid_age)
```

```python
invalid_marks= validation_df[(
    validation_df["Marks"]<0) |
    (validation_df["Marks"]>100)]
print(invalid_marks)
```

```python
valid_records=validation_df[
    (validation_df["Age"].between(16,30))&
    (validation_df["Marks"].between(0,100))
]
print(valid_records)
```

```python
validation_df["Invalid_Age"] = (
    ~validation_df["Age"].between(16, 30)
)

validation_df["Invalid_Marks"] = (
    ~validation_df["Marks"].between(0, 100)
)

print(validation_df)
```
