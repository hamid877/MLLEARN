# 11. Seaborn — Statistical Data Visualization

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

students = pd.DataFrame({
    "Name": [
        "Ali", "Sara", "John", "Aisha", "David",
        "Riya", "Ahmed", "Neha", "Rahul", "Zoya"
    ],
    "Age": [19, 20, 18, 19, 21, 20, 19, 18, 21, 20],
    "Study_Hours": [3, 5, 2, 4, 3, 6, 2, 5, 4, 7],
    "Marks": [78, 91, 45, 85, 67, 95, 50, 88, 72, 98],
    "Subject": [
        "Maths", "Physics", "Maths", "English", "Physics",
        "Maths", "English", "Physics", "Maths", "English"
    ]
})

print(students)
```

```python
sns.histplot(
    data=students,
    x="Marks",
    bins=5
)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
sns.histplot(
    data=students,
    x="Marks",
    bins=5,
    kde=True
)

plt.title("Marks Distribution with KDE")
plt.show()
```

```python
sns.boxplot(
    data=students,
    x="Marks"
)

plt.title("Distribution of Student Marks")
plt.show()
```

```python
sns.boxplot(
    data=students,
    x="Subject",
    y="Marks"
)

plt.title("Marks Distribution by Subject")
plt.show()
```

```python
correlation_matrix = students[
    ["Age", "Study_Hours", "Marks"]
].corr()

print(correlation_matrix)
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Feature Correlation Matrix")
plt.show()
```
