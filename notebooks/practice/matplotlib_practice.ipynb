```python
import matplotlib.pyplot as plt
import numpy as np

study_hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
marks = np.array([35, 40, 50, 55, 70, 85, 88, 95])

fig, ax = plt.subplots()

ax.plot(study_hours, marks)

ax.set_title("Study Hours vs Marks")
ax.set_xlabel("Study Hours")
ax.set_ylabel("Marks")

ax.grid(True)

plt.show()
```

```python
student_names = ["Ali", "Sara", "John", "Aisha", "David"]
student_marks = [78, 91, 45, 85, 67]

fig, axes = plt.subplots(
    nrows=1,
    ncols=3,
    figsize=(15, 4)
)

# Chart 1: Bar chart
axes[0].bar(student_names, student_marks)
axes[0].set_title("Student Marks")
axes[0].set_xlabel("Students")
axes[0].set_ylabel("Marks")

# Chart 2: Line chart
axes[1].plot(student_names, student_marks, marker="o")
axes[1].set_title("Marks Trend")
axes[1].set_xlabel("Students")
axes[1].set_ylabel("Marks")

# Chart 3: Histogram
axes[2].hist(student_marks, bins=5, edgecolor="black")
axes[2].set_title("Marks Distribution")
axes[2].set_xlabel("Marks")
axes[2].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
student_names = ["Ali", "Sara", "John", "Aisha", "David"]

student_marks = np.array([78, 91, 45, 85, 67])

study_hours = np.array([3, 5, 2, 4, 3])
```

```python
# Create Figure and Axes
fig, ax = plt.subplots()

# Draw a line plot
ax.plot(study_hours, student_marks)

# Display the plot
plt.show()
```

```python
# Create Figure with specific size (width=8, height=5)
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(study_hours, student_marks, marker='o')

# Set titles and labels
ax.set_title("Impact of Study Hours on Student Marks", fontsize=14)
ax.set_xlabel("Study Hours", fontsize=12)
ax.set_ylabel("Marks Obtained", fontsize=12)

plt.show()

```

```python
# Create a 2x2 grid
fig, axs = plt.subplots(2, 2, figsize=(11, 9))

# [0, 0] Top-Left: Line plot
axs[0, 0].plot(study_hours, student_marks, color='indigo', marker='s')
axs[0, 0].set_title("Line Plot: Hours vs Marks")

# [0, 1] Top-Right: Bar chart
axs[0, 1].bar(student_names, student_marks, color='teal')
axs[0, 1].set_title("Bar Chart: Student Performance")

# [1, 0] Bottom-Left: Histogram
axs[1, 0].hist(student_marks, bins=4, color='gold', edgecolor='black')
axs[1, 0].set_title("Histogram: Marks Distribution")

# [1, 1] Bottom-Right: Scatter plot
axs[1, 1].scatter(study_hours, student_marks, color='darkorange', s=80)
axs[1, 1].set_title("Scatter Plot: Correlation")

# Clean formatting and spacing
plt.tight_layout()
fig.savefig("student_analysis.png",dpi=300,bbox_inches="tight")
plt.show()

```
